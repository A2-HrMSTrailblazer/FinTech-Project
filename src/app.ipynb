```python
import os
import json
import pandas as pd
from dotenv import load_dotenv
from google import genai

# 1. Load environment variables from .env file
load_dotenv()

# 2. Initialize the Gemini client using the stable alias
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

# 3. Load the mock inventory ledger into a Pandas DataFrame
ledger_path = "../data/inventory_ledger.json"  # Adjust path if your notebook is in root vs src/
if not os.path.exists(ledger_path):
    ledger_path = "data/inventory_ledger.json" # Fallback if running from root

with open(ledger_path, "r") as f:
    inventory_data = json.load(f)

df_inventory = pd.DataFrame(inventory_data)

print("Setup Complete!")
print(f"Loaded {len(df_inventory)} inventory items from ledger.")
display(df_inventory)
```
